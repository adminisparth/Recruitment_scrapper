```python
# STEP 5 — Normalize Primary Keyword / Skill

import re

def normalize_skill(value):
    # Handle missing values
    if pd.isna(value):
        return ""

    # Convert to string and lowercase
    value = str(value).lower().strip()

    # Normalize common variations
    value = value.replace("&", "and")

    # Remove unwanted characters
    value = re.sub(r"[^a-z0-9+#.\- ]", " ", value)

    # Remove extra spaces
    value = re.sub(r"\s+", " ", value)

    return value.strip()


# Apply to jobs
jobs["skill"] = jobs["Primary Keyword"].apply(normalize_skill)

# Apply to candidates
candidates["skill"] = candidates["Primary Keyword"].apply(normalize_skill)


# Check results
print("Top Job Skills:")
print(jobs["skill"].value_counts().head(30))

print("\nTop Candidate Skills:")
print(candidates["skill"].value_counts().head(30))
```

```python
# STEP 6 — Generate positive job-candidate pairs

import numpy as np
import pandas as pd
from collections import defaultdict

# Random generator for reproducibility
rng = np.random.default_rng(42)

# ---------------------------------------------------------
# 1. Create an index of candidates grouped by skill
# ---------------------------------------------------------

candidate_by_skill = defaultdict(list)

for idx, skill in candidates["skill"].items():
    if skill != "":
        candidate_by_skill[skill].append(idx)

print("Number of skill groups:", len(candidate_by_skill))


# ---------------------------------------------------------
# 2. Generate positive pairs
# ---------------------------------------------------------

positive_pairs = []

MAX_POSITIVE_PAIRS = 100_000
CANDIDATES_PER_JOB = 3

for job_idx, job in jobs.iterrows():

    skill = job["skill"]

    # Find candidates with the same primary skill
    candidate_pool = candidate_by_skill.get(skill, [])

    if len(candidate_pool) == 0:
        continue

    # Select up to 3 candidates
    sample_size = min(
        CANDIDATES_PER_JOB,
        len(candidate_pool)
    )

    selected_candidates = rng.choice(
        candidate_pool,
        size=sample_size,
        replace=False
    )

    for candidate_idx in selected_candidates:

        positive_pairs.append({
            "job_idx": job_idx,
            "candidate_idx": candidate_idx,
            "label": 1
        })

    # Stop once we have enough
    if len(positive_pairs) >= MAX_POSITIVE_PAIRS:
        break


positive_pairs = pd.DataFrame(positive_pairs)

print("\nPositive pairs:", len(positive_pairs))
print(positive_pairs.head())
```

```python
# STEP 7 — Generate negative job-candidate pairs

negative_pairs = []

MAX_NEGATIVE_PAIRS = 100_000
CANDIDATES_PER_JOB = 3

# Convert candidate indices to numpy array for faster sampling
all_candidate_indices = candidates.index.to_numpy()

for job_idx, job in jobs.iterrows():

    job_skill = job["skill"]

    # Candidates whose primary skill is DIFFERENT
    different_skill_candidates = candidates.index[
        candidates["skill"] != job_skill
    ].to_numpy()

    if len(different_skill_candidates) == 0:
        continue

    # Select up to 3 negative candidates
    sample_size = min(
        CANDIDATES_PER_JOB,
        len(different_skill_candidates)
    )

    selected_candidates = rng.choice(
        different_skill_candidates,
        size=sample_size,
        replace=False
    )

    for candidate_idx in selected_candidates:

        negative_pairs.append({
            "job_idx": job_idx,
            "candidate_idx": candidate_idx,
            "label": 0
        })

    if len(negative_pairs) >= MAX_NEGATIVE_PAIRS:
        break


negative_pairs = pd.DataFrame(negative_pairs)

print("Negative pairs:", len(negative_pairs))
print("\nSample:")
print(negative_pairs.head())
```

```python
# Recreate the combined pairs dataframe

pairs = pd.concat(
    [positive_pairs, negative_pairs],
    ignore_index=True
)

pairs = pairs.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Total pairs:", len(pairs))
print("\nLabels:")
print(pairs["label"].value_counts())
```

```python
EMBEDDING_BATCH_SIZE = 256

def generate_embeddings(texts, batch_size=256):
    embeddings = []

    for start in tqdm(
        range(0, len(texts), batch_size),
        desc="Generating embeddings"
    ):
        batch = texts[start:start + batch_size]

        vectors = use_model(batch).numpy()

        embeddings.append(vectors)

    return np.vstack(embeddings).astype("float32")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

EMBEDDING_DIM = 512
TOWER_DIM = 256


def build_tower(name):
    inputs = keras.Input(
        shape=(EMBEDDING_DIM,),
        name=f"{name}_input"
    )

    x = layers.Dense(
        512,
        activation="relu"
    )(inputs)

    x = layers.Dropout(0.1)(x)

    x = layers.Dense(
        TOWER_DIM
    )(x)

    # Keras 3 compatible L2 normalization
    outputs = layers.UnitNormalization(
        axis=1,
        name=f"{name}_normalize"
    )(x)

    return keras.Model(
        inputs=inputs,
        outputs=outputs,
        name=name
    )


job_tower = build_tower("job_tower")
candidate_tower = build_tower("candidate_tower")

job_tower.summary()
candidate_tower.summary()
```

```python
class TwoTowerModel(keras.Model):

    def __init__(self, job_tower, candidate_tower):
        super().__init__()

        self.job_tower = job_tower
        self.candidate_tower = candidate_tower

        self.log_temperature = self.add_weight(
            name="log_temperature",
            shape=(),
            initializer=keras.initializers.Constant(
                np.log(0.07)
            ),
            trainable=True
        )

    def call(self, inputs):
        job_vectors, candidate_vectors = inputs

        job_output = self.job_tower(job_vectors)
        candidate_output = self.candidate_tower(
            candidate_vectors
        )

        temperature = tf.exp(
            self.log_temperature
        )

        similarity = tf.reduce_sum(
            job_output * candidate_output,
            axis=1
        )

        return similarity / temperature
```

```python
BATCH_SIZE = 128

train_job_idx = train_positive["job_idx"].values
train_candidate_idx = train_positive["candidate_idx"].values

val_job_idx = val_positive["job_idx"].values
val_candidate_idx = val_positive["candidate_idx"].values


def make_dataset(job_idx, candidate_idx, shuffle=True):
    job_vecs = job_embeddings[job_idx].astype("float32")
    candidate_vecs = candidate_embeddings[candidate_idx].astype("float32")

    dataset = tf.data.Dataset.from_tensor_slices(
        (job_vecs, candidate_vecs)
    )

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(job_vecs),
            reshuffle_each_iteration=True
        )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds = make_dataset(
    train_job_idx,
    train_candidate_idx,
    shuffle=True
)

val_ds = make_dataset(
    val_job_idx,
    val_candidate_idx,
    shuffle=False
)

print("Datasets created successfully!")
```

```python
def contrastive_loss(job_vectors, candidate_vectors):
    # Similarity matrix
    logits = tf.matmul(
        job_vectors,
        candidate_vectors,
        transpose_b=True
    )

    # Temperature
    temperature = tf.exp(model.log_temperature)
    logits = logits / temperature

    batch_size = tf.shape(logits)[0]

    labels = tf.range(batch_size)

    # Job -> Candidate
    loss_job_to_candidate = tf.keras.losses.sparse_categorical_crossentropy(
        labels,
        logits,
        from_logits=True
    )

    # Candidate -> Job
    loss_candidate_to_job = tf.keras.losses.sparse_categorical_crossentropy(
        labels,
        tf.transpose(logits),
        from_logits=True
    )

    return (
        tf.reduce_mean(loss_job_to_candidate) +
        tf.reduce_mean(loss_candidate_to_job)
    ) / 2
```

```python
optimizer = keras.optimizers.Adam(
    learning_rate=1e-4
)

print("Optimizer ready!")
```

```python
@tf.function
def train_step(job_batch, candidate_batch):

    with tf.GradientTape() as tape:

        job_vectors = model.job_tower(job_batch)
        candidate_vectors = model.candidate_tower(candidate_batch)

        loss = contrastive_loss(
            job_vectors,
            candidate_vectors
        )

    gradients = tape.gradient(
        loss,
        model.trainable_variables
    )

    optimizer.apply_gradients(
        zip(gradients, model.trainable_variables)
    )

    return loss
```

```python
for job_batch, candidate_batch in train_ds.take(1):

    loss = train_step(
        job_batch,
        candidate_batch
    )

    print("First batch loss:", float(loss))
```

```python
EPOCHS = 10

for epoch in range(EPOCHS):

    train_losses = []

    for job_batch, candidate_batch in train_ds:

        loss = train_step(
            job_batch,
            candidate_batch
        )

        train_losses.append(float(loss))

    avg_train_loss = np.mean(train_losses)

    print(
        f"Epoch {epoch + 1}/{EPOCHS} "
        f"- Train Loss: {avg_train_loss:.4f}"
    )
```

```python
def get_tower_embeddings(dataset):
    job_vectors = []
    candidate_vectors = []

    for job_batch, candidate_batch in dataset:
        job_vectors.append(
            model.job_tower(job_batch, training=False).numpy()
        )
        candidate_vectors.append(
            model.candidate_tower(candidate_batch, training=False).numpy()
        )

    return (
        np.vstack(job_vectors),
        np.vstack(candidate_vectors)
    )


val_job_vectors, val_candidate_vectors = get_tower_embeddings(val_ds)

print("Validation job vectors:", val_job_vectors.shape)
print("Validation candidate vectors:", val_candidate_vectors.shape)
```

```python
def evaluate_retrieval(job_vectors, candidate_vectors, k_values=[1, 5, 10]):

    # Similarity matrix
    similarities = np.matmul(
        job_vectors,
        candidate_vectors.T
    )

    ranks = []

    for i in range(len(job_vectors)):

        ranking = np.argsort(
            -similarities[i]
        )

        # Correct candidate is at the same position
        correct_position = np.where(
            ranking == i
        )[0][0]

        ranks.append(correct_position + 1)

    ranks = np.array(ranks)

    results = {}

    for k in k_values:
        results[f"Recall@{k}"] = np.mean(ranks <= k)

    results["MRR"] = np.mean(1 / ranks)

    return results


results = evaluate_retrieval(
    val_job_vectors,
    val_candidate_vectors
)

results
```

```python
def evaluate_real_retrieval(
    pairs,
    job_embeddings,
    candidate_embeddings,
    job_tower,
    candidate_tower,
    k_values=(1, 5, 10)
):
    # Unique jobs in validation set
    job_ids = pairs["job_idx"].unique()

    # Encode all candidates that appear in validation
    candidate_ids = pairs["candidate_idx"].unique()

    # Get embeddings
    job_input = job_embeddings[job_ids].astype("float32")
    candidate_input = candidate_embeddings[candidate_ids].astype("float32")

    # Two-Tower representations
    job_vectors = job_tower(
        job_input,
        training=False
    ).numpy()

    candidate_vectors = candidate_tower(
        candidate_input,
        training=False
    ).numpy()

    # Similarity matrix
    similarity = np.matmul(
        job_vectors,
        candidate_vectors.T
    )

    # candidate index -> position in matrix
    candidate_position = {
        cid: i
        for i, cid in enumerate(candidate_ids)
    }

    # Actual positive candidates for every job
    positive_pairs = pairs[pairs["label"] == 1]

    positives_by_job = (
        positive_pairs
        .groupby("job_idx")["candidate_idx"]
        .apply(set)
        .to_dict()
    )

    recalls = {k: [] for k in k_values}
    reciprocal_ranks = []

    for row_idx, job_id in enumerate(job_ids):

        if job_id not in positives_by_job:
            continue

        positive_candidates = positives_by_job[job_id]

        ranking = np.argsort(
            -similarity[row_idx]
        )

        ranked_candidate_ids = [
            candidate_ids[i]
            for i in ranking
        ]

        # Positions of all relevant candidates
        positive_ranks = [
            rank + 1
            for rank, candidate_id
            in enumerate(ranked_candidate_ids)
            if candidate_id in positive_candidates
        ]

        if not positive_ranks:
            continue

        best_rank = min(positive_ranks)

        for k in k_values:
            recalls[k].append(
                1 if best_rank <= k else 0
            )

        reciprocal_ranks.append(
            1 / best_rank
        )

    results = {}

    for k in k_values:
        results[f"Recall@{k}"] = np.mean(
            recalls[k]
        )

    results["MRR"] = np.mean(
        reciprocal_ranks
    )

    return results
```

```python
def skill_similarity(job_skills, candidate_skills):

    if not job_skills:
        return 0.0

    if not candidate_skills:
        return 0.0

    intersection = len(
        job_skills & candidate_skills
    )

    union = len(
        job_skills | candidate_skills
    )

    return intersection / union if union else 0.0
```

```python
hard_negative_rows = []

positive_train = train_pairs[
    train_pairs["label"] == 1
]

for _, row in positive_train.iterrows():

    job_idx = int(row["job_idx"])
    positive_candidate = int(row["candidate_idx"])

    job_skills = job_skill_map.get(
        job_idx,
        set()
    )

    if not job_skills:
        continue

    # Candidates sharing at least one skill
    candidate_pool = set()

    for skill in job_skills:
        candidate_pool.update(
            skill_to_candidates.get(skill, set())
        )

    # Remove the positive candidate
    candidate_pool.discard(
        positive_candidate
    )

    scored_candidates = []

    for candidate_idx in candidate_pool:

        candidate_skills = candidate_skill_map.get(
            candidate_idx,
            set()
        )

        similarity = skill_similarity(
            job_skills,
            candidate_skills
        )

        # Keep plausible candidates
        if similarity > 0:
            scored_candidates.append(
                (candidate_idx, similarity)
            )

    # Highest-overlap candidates = hardest negatives
    scored_candidates.sort(
        key=lambda x: x[1],
        reverse=True
    )

    # Take up to 2 hard negatives
    for candidate_idx, similarity in scored_candidates[:2]:

        hard_negative_rows.append({
            "job_idx": job_idx,
            "candidate_idx": candidate_idx,
            "label": 0,
            "skill_similarity": similarity
        })
```

```python
@tf.function
def ranking_train_step(
    job_batch,
    positive_batch,
    negative_batch
):

    with tf.GradientTape() as tape:

        # Job representation
        job_vector = model.job_tower(
            job_batch,
            training=True
        )

        # Positive candidate representation
        positive_vector = model.candidate_tower(
            positive_batch,
            training=True
        )

        # Negative candidate representation
        negative_vector = model.candidate_tower(
            negative_batch,
            training=True
        )

        # Cosine similarity because vectors are normalized
        positive_score = tf.reduce_sum(
            job_vector * positive_vector,
            axis=1
        )

        negative_score = tf.reduce_sum(
            job_vector * negative_vector,
            axis=1
        )

        loss = ranking_loss(
            positive_score,
            negative_score
        )

    gradients = tape.gradient(
        loss,
        model.trainable_variables
    )

    optimizer.apply_gradients(
        zip(
            gradients,
            model.trainable_variables
        )
    )

    return loss
```

```python
def parse_years(value):
    if pd.isna(value):
        return np.nan

    text = str(value).lower()

    # Examples: 3, 3.5, "3 years", "2-4"
    nums = re.findall(r"\d+(?:\.\d+)?", text)

    if not nums:
        return np.nan

    nums = [float(x) for x in nums]

    if len(nums) >= 2:
        return sum(nums[:2]) / 2

    return nums[0]
```

```python
def get_candidate_pool(job_idx, max_pool=100):
    job_skills = jobs.iloc[job_idx]["skills"]

    pool = set()

    for skill in job_skills:
        for candidate_idx in candidate_skill_index.get(skill, []):
            pool.add(candidate_idx)

            if len(pool) >= max_pool:
                return list(pool)

    return list(pool)
```

```python
def skill_score(job_skills, candidate_skills):
    if not job_skills:
        return 0.0

    if not candidate_skills:
        return 0.0

    overlap = len(job_skills & candidate_skills)

    return overlap / len(job_skills)
```

```python
def experience_score(job_exp, candidate_exp):
    if pd.isna(job_exp) or pd.isna(candidate_exp):
        return 0.5

    difference = abs(job_exp - candidate_exp)

    if difference <= 0.5:
        return 1.0
    elif difference <= 1:
        return 0.9
    elif difference <= 2:
        return 0.7
    elif difference <= 3:
        return 0.4
    else:
        return 0.1
```

```python
def semantic_score(job_idx, candidate_idx):
    j = job_embeddings[job_idx]
    c = candidate_embeddings[candidate_idx]

    score = np.dot(j, c)

    return float(np.clip(score, 0, 1))
```

```python
def calculate_relevance(job_idx, candidate_idx):

    job = jobs.iloc[job_idx]
    candidate = candidates.iloc[candidate_idx]

    skill = skill_score(
        job["skills"],
        candidate["skills"]
    )

    experience = experience_score(
        job["experience"],
        candidate["experience"]
    )

    semantic = semantic_score(
        job_idx,
        candidate_idx
    )

    pos_similarity = cosine_similarity(
        job_position_vectors[job_idx],
        candidate_position_vectors[candidate_idx]
    )[0][0]

    score = (
        0.40 * skill +
        0.30 * semantic +
        0.20 * experience +
        0.10 * pos_similarity
    )

    return float(score)
```
